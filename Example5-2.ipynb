```python
from tqdm import tqdm
import torch
from cibsde import nn, utils
import matplotlib.pyplot as plt
device = 'cpu'
seed = 666
torch.manual_seed(seed)
```

## Parameter settings

```python
# model parameters
num_sphere = 3
t = torch.tensor(1.,device=device)
D = torch.tensor(0.1,device=device)
r = torch.tensor([1,1.3,1.5],device=device)
R = 2
N = 100
# training parameters
batch = 512
epoch = 10**4
lr = 1e-3
train_params = {
    'epoch': epoch,
    'batch': batch,
    'lr': lr
}
```

## Model initialization

```python
def data_gen(batch):
    u1 = torch.rand([batch,num_sphere],device=device)
    u2 = torch.rand([batch,num_sphere],device=device)*(1-r/R/2)
    phi = 2 * torch.pi * u1
    theta = torch.arccos(1 - u2)
    theta[:,:-1] = 0
    x = torch.sin(theta) * torch.cos(phi)
    y = torch.sin(theta) * torch.sin(phi)
    z = torch.cos(theta)
    return torch.stack([x,y,z],dim=2).reshape([batch,num_sphere*3]).to(device)

def pc(i,x):
    x = x.reshape([batch*num_sphere,3])
    x_polar = utils.polar_corr(x).reshape([batch*num_sphere,2])
    return (x_polar[:,0]<torch.pi/4).reshape([batch,num_sphere]).prod(dim=1,keepdim=True).float()

def xb(x):
    thetab = torch.arccos(r/R/2)
    x = x.reshape([batch*num_sphere,3])
    x_polar = utils.polar_corr(x).reshape([batch,num_sphere,2])
    in_ornot = (x_polar[:,:,0] < thetab).reshape([batch*num_sphere])
    xb_ = x_polar.clone()
    xb_[:,:,0] = thetab
    xb_ = utils.euclid_corr(xb_.reshape([batch*num_sphere,2]))
    nb_ = torch.cat([xb_[:,:1]*xb_[:,2:],xb_[:,1:2]*xb_[:,2:],-xb_[:,:2].norm(dim=1,keepdim=True)**2],dim=1)
    nb_ = nb_ / nb_.norm(dim=1,keepdim=True)
    return xb_, nb_, ~in_ornot

def hitb(x):
    return torch.zeros(batch,device=device).bool()

def f(t,xt,p):
    return torch.zeros([batch,1],device=device)

def h(t,x):
    return torch.zeros([batch,1],device=device)

model = nn.SphereIBSDE(
    dx = num_sphere,
    t = t,
    f = f,
    D = D/r**2,
    h = h,
    pc = pc,
    data_gen = data_gen,
    N = N,
    refb = xb,
    hitb = hitb
).to(device)
```

## Loading reference solution

```python
y_rel = torch.load('./datasets/example5-2.pth')
```

## Model training

```python
def train(model, params:dict):
    epoch = params['epoch']
    batch = params['batch']
    lr = params['lr']
    theta = torch.linspace(0,torch.arccos(r[-1]/R/2),101)
    phi = torch.linspace(0,torch.pi*2,200)
    theta,phi = torch.meshgrid(theta[:100],phi,indexing='ij')
    x_polar = torch.stack([theta.reshape(100*200),phi.reshape(100*200)],dim=1)
    x = utils.euclid_corr(x_polar).to(device)
    xx = torch.tensor([0,0,1,0,0,1]).expand([100*200,6]).to(device)
    x = torch.cat([xx,x],dim=1)

    optim = torch.optim.Adam(model.parameters(),lr=lr)
    loss_fun = torch.nn.MSELoss()
    
    loss_values = torch.zeros(epoch)
    rel_error = torch.zeros(epoch)
    pbar = tqdm(range(epoch))
    for i in pbar:
        model.train()
        optim.zero_grad()
        u_pre, u_rel = model(batch)
        loss = loss_fun(u_pre,u_rel)
        loss.backward()
        optim.step()

        model.eval()
        loss_values[i] = loss.item()
        with torch.no_grad():
            pre = model.p(x).reshape([100,200]).detach()
            rel_error[i] = torch.sqrt(((pre - y_rel)**2).mean()/((y_rel**2).mean()))
        pbar.set_postfix(loss='%e' % loss_values[i], error='%e' % rel_error[i])
    print("\nTraining has been completed.")
    return loss_values, rel_error
```

```python
loss_value, rel_error = train(
    model,
    train_params
)
```

## PINN

```python
class PINN(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.p = utils.FNN(3*num_sphere+1,1)
        self.r = r
        self.num = num_sphere

    def forward(self, batch):
        u1 = torch.rand([batch,self.num],device=device)
        u2 = torch.rand([batch,self.num],device=device)*(1-self.r/R/2)

        phi = 2 * torch.pi * u1
        theta = torch.arccos(1 - u2)
        phi = phi[(theta>1e-4).all(dim=1)]
        theta = theta[(theta>1e-4).all(dim=1)]

        phi.requires_grad = True
        theta.requires_grad = True

        x = torch.sin(theta) * torch.cos(phi)
        y = torch.sin(theta) * torch.sin(phi)
        z = torch.cos(theta)
        ps = torch.stack([x, y, z], dim=2).reshape([theta.shape[0],self.num*3])
        t = torch.rand([theta.shape[0],1],device=device,requires_grad=True)

        phib = 2 * torch.pi * torch.rand([batch,self.num],device=device)
        thetab = torch.arccos(1 - torch.rand([batch,self.num],device=device)*(1-self.r/R/2))

        thetab1 = torch.ones([batch,1],device=device) * torch.arccos(self.r[0]/R/2)
        thetab1.requires_grad = True
        thetab2 = torch.ones([batch,1],device=device) * torch.arccos(self.r[1]/R/2)
        thetab2.requires_grad = True
        thetab3 = torch.ones([batch,1],device=device) * torch.arccos(self.r[2]/R/2)
        thetab3.requires_grad = True

        thetabb1 = torch.cat([thetab1,thetab[:,1:]],dim=1)
        thetabb2 = torch.cat([thetab[:,:1],thetab2,thetab[:,2:]],dim=1)
        thetabb3 = torch.cat([thetab[:,:2],thetab3],dim=1)
        thetabb = torch.cat([thetab1,thetab2,thetab3],dim=1)

        xb1 = torch.sin(thetabb1) * torch.cos(phib)
        yb1 = torch.sin(thetabb1) * torch.sin(phib)
        zb1 = torch.cos(thetabb1)
        pbs1 = torch.stack([xb1, yb1, zb1], dim=2).reshape([batch,self.num*3])

        xb2 = torch.sin(thetabb2) * torch.cos(phib)
        yb2 = torch.sin(thetabb2) * torch.sin(phib)
        zb2 = torch.cos(thetabb2)
        pbs2 = torch.stack([xb2, yb2, zb2], dim=2).reshape([batch,self.num*3])

        xb3 = torch.sin(thetabb3) * torch.cos(phib)
        yb3 = torch.sin(thetabb3) * torch.sin(phib)
        zb3 = torch.cos(thetabb3)
        pbs3 = torch.stack([xb3, yb3, zb3], dim=2).reshape([batch,self.num*3])

        xbb = torch.sin(thetabb) * torch.cos(phib)
        ybb = torch.sin(thetabb) * torch.sin(phib)
        zbb = torch.cos(thetabb)
        pbsb = torch.stack([xbb, ybb, zbb], dim=2).reshape([batch,self.num*3])

        tb = torch.rand([batch,1],device=device,requires_grad=True)

        u1 = torch.rand([batch,self.num],device=device)
        u2 = torch.rand([batch,self.num],device=device)*(1-self.r/R/2)

        phiT = 2 * torch.pi * u1
        thetaT = torch.arccos(1 - u2)

        xT = torch.sin(thetaT) * torch.cos(phiT)
        yT = torch.sin(thetaT) * torch.sin(phiT)
        zT = torch.cos(thetaT)
        psT = torch.stack([xT, yT, zT], dim=2).reshape([batch,self.num*3])
        T = torch.ones([batch,1],device=device)

        sintheta = torch.sin(theta)
        u = self.p(torch.cat([t,ps],dim=1))
        grad_theta = torch.autograd.grad(u,theta,torch.ones_like(u),retain_graph=True,create_graph=True)[0]
        grad_phi = torch.autograd.grad(u,phi,torch.ones_like(u),retain_graph=True,create_graph=True)[0]
        
        grad_sintheta = grad_theta*sintheta
        singrad_theta1 = torch.autograd.grad(grad_sintheta[:,:1],theta,torch.ones_like(grad_sintheta[:,:1]),retain_graph=True,create_graph=True)[0][:,:1]
        singrad_theta2 = torch.autograd.grad(grad_sintheta[:,1:2],theta,torch.ones_like(grad_sintheta[:,1:2]),retain_graph=True,create_graph=True)[0][:,1:2]
        singrad_theta3 = torch.autograd.grad(grad_sintheta[:,2:],theta,torch.ones_like(grad_sintheta[:,2:]),retain_graph=True,create_graph=True)[0][:,2:]
        singrad_theta = torch.cat([singrad_theta1,singrad_theta2,singrad_theta3],dim=1)
        
        grad2_phi1 = torch.autograd.grad(grad_phi[:,:1],phi,torch.ones_like(grad_phi[:,:1]),retain_graph=True,create_graph=True)[0][:,:1]
        grad2_phi2 = torch.autograd.grad(grad_phi[:,1:2],phi,torch.ones_like(grad_phi[:,1:2]),retain_graph=True,create_graph=True)[0][:,1:2]
        grad2_phi3 = torch.autograd.grad(grad_phi[:,2:],phi,torch.ones_like(grad_phi[:,2:]),retain_graph=True,create_graph=True)[0][:,2:]
        grad2_phi = torch.cat([grad2_phi1,grad2_phi2,grad2_phi3],dim=1)
        ub1 = self.p(torch.cat([tb,pbs1],dim=1))
        grad_ub1 = torch.autograd.grad(ub1,thetab1,torch.ones_like(ub1),retain_graph=True,create_graph=True)[0]
        ub2 = self.p(torch.cat([tb,pbs2],dim=1))
        grad_ub2 = torch.autograd.grad(ub2,thetab2,torch.ones_like(ub2),retain_graph=True,create_graph=True)[0]
        ub3 = self.p(torch.cat([tb,pbs3],dim=1))
        grad_ub3 = torch.autograd.grad(ub3,thetab3,torch.ones_like(ub3),retain_graph=True,create_graph=True)[0]
        ubb = self.p(torch.cat([tb,pbsb],dim=1))
        grad_ubb1 = torch.autograd.grad(ubb,thetab1,torch.ones_like(ubb),retain_graph=True,create_graph=True)[0]
        grad_ubb2 = torch.autograd.grad(ubb,thetab2,torch.ones_like(ubb),retain_graph=True,create_graph=True)[0]
        grad_ubb3 = torch.autograd.grad(ubb,thetab3,torch.ones_like(ubb),retain_graph=True,create_graph=True)[0]
        grad_ubb = torch.cat([grad_ub1,grad_ub2,grad_ub3,grad_ubb1,grad_ubb2,grad_ubb3])
        
        ut = torch.autograd.grad(u,t,torch.ones_like(u),retain_graph=True,create_graph=True)[0]
        uT = self.p(torch.cat([T,psT],dim=1))
        
        return torch.cat([ut,grad_ubb,uT]), torch.cat([-(D*(singrad_theta/sintheta+grad2_phi/sintheta**2)/self.r**2).sum(dim=1,keepdim=True),torch.zeros_like(grad_ubb),pc(100,psT)])
```

## PINN initialization

```python
pinn = PINN().to(device)
```

## PINN training

```python
def train(model, params:dict):
    epoch = params['epoch']
    batch = params['batch']
    lr = params['lr']
    theta = torch.linspace(0,torch.arccos(r[-1]/R/2),101)
    phi = torch.linspace(0,torch.pi*2,200)
    theta,phi = torch.meshgrid(theta[:100],phi,indexing='ij')
    x_polar = torch.stack([theta.reshape(100*200),phi.reshape(100*200)],dim=1)
    x = utils.euclid_corr(x_polar).to(device)
    xx = torch.tensor([0,0,1,0,0,1.]).expand([100*200,6]).to(device)
    x = torch.cat([xx,x],dim=1)
    t = torch.ones([100*200,1],device=device)*0

    optim = torch.optim.Adam(model.parameters(),lr=lr)
    loss_fun = torch.nn.MSELoss()
    
    loss_values = torch.zeros(epoch)
    rel_error = torch.zeros(epoch)
    pbar = tqdm(range(epoch))
    for i in pbar:
        model.train()
        optim.zero_grad()
        u_pre, u_rel = model(batch)
        loss = loss_fun(u_pre,u_rel)
        loss.backward()
        optim.step()

        model.eval()
        loss_values[i] = loss.item()
        with torch.no_grad():
            pre = model.p(torch.cat([t,x],dim=1)).reshape([100,200]).detach().to(device)
            rel_error[i] = torch.sqrt(((pre - y_rel.to(device))**2).sum()/((y_rel.to(device)**2).sum()))
        pbar.set_postfix(loss='%e' % loss_values[i], error='%e' % rel_error[i])
    print("\nTraining has been completed.")
    return loss_values, rel_error
```

```python
loss_values_pinn, rel_error_pinn = train(
    pinn,
    train_params
)
```
